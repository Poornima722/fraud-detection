```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

thresholds = [
    0.50, 0.60, 0.70, 0.75, 0.80,
    0.85, 0.90, 0.92, 0.94, 0.96, 0.98
]

for threshold in thresholds:
    y_val_pred = (y_val_prob >= threshold).astype(int)

    precision = precision_score(
        y_val, y_val_pred, zero_division=0
    )

    recall = recall_score(
        y_val, y_val_pred, zero_division=0
    )

    f1 = f1_score(
        y_val, y_val_pred, zero_division=0
    )

    tn, fp, fn, tp = confusion_matrix(
        y_val, y_val_pred
    ).ravel()

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.4f} | "
        f"Recall: {recall:.4f} | "
        f"F1: {f1:.4f} | "
        f"FP: {fp} | FN: {fn}"
    )
```

```python
import numpy as np

thresholds_cost = np.arange(0.10, 1.00, 0.01)

scenarios = {
    "Balanced (1:1)": (1, 1),
    "Fraud 5x costly (1:5)": (1, 5),
    "Fraud 10x costly (1:10)": (1, 10)
}

for scenario_name, (fp_cost, fn_cost) in scenarios.items():

    best_threshold = None
    lowest_cost = float("inf")

    for threshold in thresholds_cost:

        y_val_pred = (y_val_prob >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_val, y_val_pred
        ).ravel()

        total_cost = (
            fp * fp_cost
            +
            fn * fn_cost
        )

        if total_cost < lowest_cost:
            lowest_cost = total_cost
            best_threshold = threshold

    print(
        f"{scenario_name}: "
        f"Best threshold = {best_threshold:.2f}, "
        f"Minimum cost = {lowest_cost:.0f}"
    )
```

```python
negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

final_fraud_weight = negative / positive

print("Legitimate:", negative)
print("Fraud:", positive)
print("Final fraud class weight:", final_fraud_weight)
```

```python
from catboost import CatBoostClassifier

final_model = CatBoostClassifier(
    iterations=100,
    learning_rate=0.1,
    depth=6,
    loss_function="Logloss",
    eval_metric="PRAUC",
    cat_features=categorical_features,
    class_weights=[1, final_fraud_weight],
    random_seed=42,
    verbose=20,
    thread_count=-1
)
```

```python
final_model.fit(
    X_train,
    y_train
)
```

```python
y_test_prob = final_model.predict_proba(X_test)[:, 1]

print("Minimum probability:", y_test_prob.min())
print("Maximum probability:", y_test_prob.max())
print("Mean probability:", y_test_prob.mean())
```

```python
final_threshold = 0.78

y_test_pred = (y_test_prob >= final_threshold).astype(int)

print("Predicted fraud:", y_test_pred.sum())
print("Predicted legitimate:", (y_test_pred == 0).sum())
```

```python
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    average_precision_score
)

# Final PR-AUC
test_pr_auc = average_precision_score(y_test, y_test_prob)

# Final confusion matrix
tn, fp, fn, tp = confusion_matrix(
    y_test,
    y_test_pred
).ravel()

print("FINAL TEST RESULTS")
print("=" * 50)

print(f"PR-AUC: {test_pr_auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_pred))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_test_pred,
        digits=4
    )
)

print("\nDetailed counts:")
print("True Negatives :", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives  :", tp)
```

```python
feature_importance = final_model.get_feature_importance()

importance_df = pd.DataFrame({
    "feature": model_features,
    "importance": feature_importance
}).sort_values(
    "importance",
    ascending=False
)

print(importance_df.to_string(index=False))
```

```python
print("\nTop 10 features:")
print(
    importance_df.head(10).to_string(index=False)
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_test_pred,
    display_labels=["Legitimate", "Fraud"],
    values_format="d"
)

plt.title("Confusion Matrix - Final CatBoost Model")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds_pr = precision_recall_curve(
    y_test,
    y_test_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall,
    precision,
    label=f"PR-AUC = {test_pr_auc:.4f}"
)

# Mark our chosen threshold
threshold_index = np.argmin(
    np.abs(thresholds_pr - final_threshold)
)

plt.scatter(
    recall[threshold_index],
    precision[threshold_index],
    s=100,
    label=f"Threshold = {final_threshold}"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision–Recall Curve - Final CatBoost Model")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
